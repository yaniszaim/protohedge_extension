```python
import os
import pickle

from deephedging.run_train_torch import _build_bs_prototype_payload
from deephedging.world_torch import SimpleWorld_Spot_ATM

os.makedirs("prototypes_storage", exist_ok=True)

n_prototypes = 100

world = SimpleWorld_Spot_ATM(
    {
        "samples": 10000,
        "steps": 20,
        "black_scholes": True,
        "seed": 1234,
    }
)

payload = _build_bs_prototype_payload(world, n_prototypes)

with open("prototypes_storage/prototypes_100.pkl", "wb") as f:
    pickle.dump(payload, f)

print("saved prototypes", payload["prototypes"].shape)
```
